Hva skjer med $f(x)$ når $x$ nærmer seg et tall, men aldri når helt fram? Det er spørsmålet en **grenseverdi** svarer på. Vi undersøker det på tre måter: med en verditabell, med en graf og med sympy.

## Verditabell

Funksjonen $f(x) = \dfrac{x^2 - 9}{x - 3}$ er ikke definert for $x = 3$. Men hva skjer *nær* 3?

```python
def f(x):
    return (x**2 - 9) / (x - 3)

for h in [0.1, 0.01, 0.001, 0.0001]:
    print(f"f(3 - {h}) = {f(3 - h):.6f}    f(3 + {h}) = {f(3 + h):.6f}")
```

Begge sider nærmer seg 6. Vi skriver $\displaystyle\lim_{x \to 3} f(x) = 6$.

## Grenseverdier med sympy

`sp.limit(uttrykk, x, a)` gir grenseverdien. Legg til `'-'` eller `'+'` for å få grenseverdien fra venstre eller høyre. Uendelig skrives `sp.oo`.

```python
import sympy as sp
x = sp.symbols('x')

print(sp.limit((x**2 - 9)/(x - 3), x, 3))
print(sp.limit(1/x, x, 0, '-'), sp.limit(1/x, x, 0, '+'))
print(sp.limit((3*x + 1)/(x - 2), x, sp.oo))
```

## Kontinuitet

En funksjon er **kontinuerlig** i $x = a$ hvis

$$\lim_{x \to a^-} f(x) = \lim_{x \to a^+} f(x) = f(a)$$

Grafisk: du kan tegne grafen gjennom punktet uten å løfte blyanten.

Delt definerte funksjoner lages med `sp.Piecewise`:

```python
g = sp.Piecewise((x + 1, x < 1), (3 - x, True))     # True betyr «ellers»
print(sp.limit(g, x, 1, '-'), sp.limit(g, x, 1, '+'), g.subs(x, 1))
```

```python
# Kjør denne cellen for å sjekke svarene dine. Du trenger ikke forstå koden.
def naer(a, b, rel=1e-3):
    """Sant hvis tallet a er innenfor rel (relativt) av b."""
    try:
        a, b = float(a), float(b)
    except (TypeError, ValueError):
        return False
    if b == 0:
        return abs(a) < 1e-9
    return abs(a - b) <= rel * abs(b)


def S(tekst):
    """Gjør tekst om til et sympy-uttrykk, f.eks. S("x**2 - 1")."""
    import sympy
    return sympy.sympify(tekst, locals={n: sympy.Symbol(n) for n in ("N", "O", "Q", "S")})


def lik(a, b):
    """Sant hvis a og b er matematisk like.

    Lister og mengder sammenlignes uten hensyn til rekkefølge, tupler (koordinater) i rekkefølge.
    """
    import sympy
    if a is None:
        return False
    if isinstance(b, tuple):
        if not isinstance(a, (tuple, list, sympy.Tuple)) or len(a) != len(b):
            return False
        return all(lik(x, y) for x, y in zip(a, b))
    if isinstance(b, (list, set)):
        if isinstance(a, sympy.FiniteSet):
            a = list(a)
        if not isinstance(a, (list, tuple, set)) or len(a) != len(b):
            return False
        rest = list(a)
        for y in b:
            for i, x in enumerate(rest):
                if lik(x, y):
                    del rest[i]
                    break
            else:
                return False
        return True
    if isinstance(b, dict):
        if not isinstance(a, dict) or len(a) != len(b):
            return False
        a = {str(k): v for k, v in a.items()}
        return all(str(k) in a and lik(a[str(k)], v) for k, v in b.items())
    try:
        a, b = sympy.sympify(a), sympy.sympify(b)
    except (sympy.SympifyError, TypeError):
        return False
    vanlige = lambda u: u.subs({s: sympy.Symbol(s.name) for s in u.free_symbols})
    a, b = vanlige(a), vanlige(b)
    if isinstance(a, sympy.Set) or isinstance(b, sympy.Set):
        return a == b
    if getattr(b, "is_Relational", False):
        return a == b
    try:
        return sympy.simplify(a - b) == 0
    except TypeError:
        return a == b

SJEKKER = [
    ('1_hull', 'numerisk_grense((x² - 9)/(x - 3), 3) skal gi ca. 6', 'naer(numerisk_grense(lambda x: (x**2 - 9)/(x - 3), 3), 6, rel=1e-5)'),
    ('1_annen', 'numerisk_grense((x³ - 1)/(x - 1), 1) skal gi ca. 3', 'naer(numerisk_grense(lambda x: (x**3 - 1)/(x - 1), 1), 3, rel=1e-5)'),
    ('2a', 'g_2a skal være 1', 'lik(g_2a, 1)'),
    ('2b', 'g_2b skal være 2/5', "lik(g_2b, S('2/5'))"),
    ('2c', 'g_2c skal være 1/2', "lik(g_2c, S('1/2'))"),
    ('3', 'venstre og hoyre skal være -1 og 1, og grensen finnes ikke', 'lik(venstre, -1) and lik(hoyre, 1) and grensen_finnes == False'),
    ('4', 'a_kontinuerlig skal være 3', 'lik(a_kontinuerlig, 3)'),
    ('5_ja', 'er_kontinuerlig(x² + 1, 3x - 1, 2) skal gi True', "er_kontinuerlig(S('x**2 + 1'), S('3*x - 1'), 2) == True"),
    ('5_nei', 'er_kontinuerlig(x², x + 1, 1) skal gi False', "er_kontinuerlig(S('x**2'), S('x + 1'), 1) == False and er_kontinuerlig(S('x'), S('2*x - 1'), 1) == True"),
]


def test_alt():
    feil = 0
    for _, beskrivelse, uttrykk in SJEKKER:
        try:
            ok = eval(uttrykk)
        except Exception as e:
            ok = False
            beskrivelse += f" (krasjet: {type(e).__name__}: {e})"
        if not ok:
            print("❌", beskrivelse)
            feil += 1
    if feil == 0:
        print(f"✅ Alle testene bestått! ({len(SJEKKER)} av {len(SJEKKER)})")
    else:
        print(f"\n{len(SJEKKER) - feil} av {len(SJEKKER)} tester bestått.")


test_alt()
```
